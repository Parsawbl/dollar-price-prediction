```python
import pandas as pd
```

```python
df = pd.read_csv('../data/processed/dollar_price_final.csv', index_col=0, parse_dates=True)
df.index.name = 'Date'
df.head()
```

# ساخت فیچرهای Lag (تاخیری)

```python
df['Close_Lag_1'] = df['Close'].shift(1)
df['Close_Lag_3'] = df['Close'].shift(3)
df['Close_Lag_7'] = df['Close'].shift(7)
```

```python
df[['Close', 'Close_Lag_1', 'Close_Lag_3', 'Close_Lag_7']].head(10)
```

# Moving Average

```python
df['MA_7'] = df['Close'].rolling(window=7).mean()
df['MA_30'] = df['Close'].rolling(window=30).mean()
```

```python
df[['Close', 'MA_7', 'MA_30']].tail(10)
```

حذف ردیف‌های ناقص (به‌خاطر NaN های ابتدای داده)

```python
df = df.dropna()
```

میانگین متحرک رو یک روز عقب بکشیم

```python
df['MA_7'] = df['Close'].shift(1).rolling(window=7).mean()
df['MA_30'] = df['Close'].shift(1).rolling(window=30).mean()
```

```python
test = pd.Series([10, 20, 30, 40, 50, 60, 70])
print(test.shift(1).rolling(window=3).mean())
```
